# La Neurona Artificial

En el módulo anterior vimos que un modelo recibe **features** y produce una predicción.

Ahora abriremos por primera vez la “caja” del modelo.

La unidad básica de una red neuronal es una **neurona artificial**.

La idea central es sencilla:

```text
Features
   │
   ▼
Multiplicar por pesos
   │
   ▼
Sumar
   │
   ▼
Añadir bias
   │
   ▼
Aplicar activación
   │
   ▼
Salida
```

## 1. De una medición científica a una entrada

Imaginemos que estamos estudiando condiciones atmosféricas.

Para una observación tenemos:

- temperatura;
- humedad;
- presión atmosférica.

Podemos llamar a estas features:

\[
x_1,\;x_2,\;x_3.
\]

Una neurona recibe esos valores como entradas.

```text
Temperatura ─────► x₁ ──┐
Humedad ─────────► x₂ ──┼──► Neurona ───► salida
Presión ─────────► x₃ ──┘
```

Cada entrada se multiplica por un número llamado **weight**.


## 2. La ecuación de una neurona

Una neurona calcula primero:

\[
z = w_1x_1+w_2x_2+w_3x_3+b
\]

donde:

- \(x_i\): features;
- \(w_i\): weights;
- \(b\): bias;
- \(z\): combinación lineal de las entradas.

Después, normalmente aplicamos una función de activación:

\[
a = f(z).
\]

En este notebook nos concentraremos primero en entender \(z\).


### Ejemplo numérico

Supongamos que normalizamos tres mediciones científicas y obtenemos:

\[
x_1=0.8,\qquad x_2=0.6,\qquad x_3=0.2.
\]

Y utilizamos:

\[
w_1=0.5,\qquad w_2=-0.3,\qquad w_3=0.8,\qquad b=0.1.
\]

Calcula primero el valor de \(z\) a mano.

<details>
<summary><strong>💡 Pista</strong></summary>

Sustituye directamente:

\[
z=(0.5)(0.8)+(-0.3)(0.6)+(0.8)(0.2)+0.1
\]

</details>


````python
x1 = 0.8
x2 = 0.6
x3 = 0.2

w1 = 0.5
w2 = -0.3
w3 = 0.8

b = 0.1

z = w1*x1 + w2*x2 + w3*x3 + b

print("z =", z)
````

<details>
<summary><strong> Mostrar solución</strong></summary>

\[
z=0.48
\]

La neurona combina las tres features en un solo número.

Lo importante es que cada feature contribuye de manera diferente dependiendo de su **weight**.

</details>


## 3. ¿Qué hace un weight?

Un weight controla cuánto influye una feature en la salida.

Por ejemplo:

```text
weight positivo grande   → aumenta z cuando aumenta x
weight negativo          → reduce z cuando aumenta x
weight cercano a cero    → poca influencia
```

Vamos a verlo interactuando con una sola feature.


````python
x = 0.8

for w in [-2.0, -1.0, 0.0, 0.5, 1.0, 2.0]:
    z = w * x
    print(f"w = {w:>4}  ->  z = {z:>5.2f}")
````

### Pregunta

¿Qué ocurre cuando el weight cambia de positivo a negativo?

<details>
<summary><strong>Pista</strong></summary>

Observa el signo de \(z\).

</details>

<details>
<summary><strong>Mostrar solución</strong></summary>

Un weight positivo hace que una feature positiva aumente \(z\).

Un weight negativo hace que la misma feature reduzca \(z\).

Por eso los weights pueden aprender si una feature debe contribuir positiva o negativamente a una decisión.

</details>


## 4. ¿Qué hace el bias?

El **bias** añade un desplazamiento adicional:

\[
z=wx+b.
\]

Incluso si \(x=0\), el bias puede producir una salida distinta de cero.

Prueba varios valores.


````python
x = 0.0
w = 1.0

for b in [-2.0, -1.0, 0.0, 1.0, 2.0]:
    z = w*x + b
    print(f"b = {b:>4}  ->  z = {z:>5.2f}")
````

### Intuición

El bias permite desplazar la frontera de decisión.

Sin bias:

\[
z=wx
\]

la transición ocurre alrededor de \(x=0\).

Con bias:

\[
z=wx+b
\]

podemos mover esa transición.

Esto será muy importante cuando una red tenga que aprender fronteras a partir de datos reales.


## 5. La misma neurona con NumPy

En redes neuronales reales no escribimos:

```python
w1*x1 + w2*x2 + w3*x3
```

una y otra vez.

Usamos vectores.

\[
\mathbf{x}
=
\begin{bmatrix}
x_1\\
x_2\\
x_3
\end{bmatrix},
\qquad
\mathbf{w}
=
\begin{bmatrix}
w_1\\
w_2\\
w_3
\end{bmatrix}.
\]

Entonces:

\[
z=\mathbf{w}\cdot\mathbf{x}+b.
\]


````python
import numpy as np

x = np.array([0.8, 0.6, 0.2])
w = np.array([0.5, -0.3, 0.8])
b = 0.1

z = np.dot(w, x) + b

print("x =", x)
print("w =", w)
print("z =", z)
````
