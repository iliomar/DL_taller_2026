# Funciones de Activación

Ya sabemos que una neurona calcula:

\[
z=\mathbf{w}\cdot\mathbf{x}+b.
\]

Ahora añadimos una función:

\[
a=f(z).
\]

Esta función se llama **función de activación**.

Su papel es fundamental: permite que una red neuronal aprenda relaciones **no lineales**.

## Objetivos

Al finalizar podrás:

- entender por qué necesitamos activaciones;
- visualizar ReLU, Sigmoid y Tanh;
- interpretar su comportamiento;
- aplicar una activación a la salida de una neurona;
- reconocer qué activaciones se usan con frecuencia en capas ocultas y salidas.


````python
import numpy as np
import matplotlib.pyplot as plt
````

## 1. ¿Por qué no basta con una suma lineal?

Supongamos que una neurona calcula:

\[
z=wx+b.
\]

Si conectamos muchas transformaciones lineales sin activaciones, matemáticamente pueden colapsar en otra transformación lineal.

Por eso necesitamos una operación no lineal.

```text
Linear operation
      ↓
Activation
      ↓
Linear operation
      ↓
Activation
```

Eso permite crear fronteras y patrones mucho más complejos.


## 2. ReLU

Una de las activaciones más utilizadas es:

\[
\mathrm{ReLU}(z)=\max(0,z).
\]

Esto significa:

```text
si z < 0  →  0
si z > 0  →  z
```


````python
def relu(z):
    return np.maximum(0, z)

z = np.linspace(-5, 5, 300)

plt.figure(figsize=(7, 4))
plt.plot(z, relu(z))
plt.axhline(0, linewidth=0.8)
plt.axvline(0, linewidth=0.8)
plt.xlabel("z")
plt.ylabel("ReLU(z)")
plt.title("Función ReLU")
plt.show()
````

### Pregunta

¿Qué ocurre con valores negativos?

<details>
<summary><strong>Mostrar solución</strong></summary>

Todos los valores negativos se convierten en cero.

Los positivos pasan prácticamente sin cambio.

ReLU es muy popular en **hidden layers** porque es sencilla y suele funcionar bien.

</details>


## 3. Sigmoid

La función Sigmoid es:

\[
\sigma(z)=\frac{1}{1+e^{-z}}.
\]

Su salida siempre está entre:

\[
0 \le \sigma(z) \le 1.
\]

Eso hace que sea intuitiva para problemas binarios.


````python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

z = np.linspace(-8, 8, 300)

plt.figure(figsize=(7, 4))
plt.plot(z, sigmoid(z))
plt.axhline(0.5, linestyle="--", linewidth=0.8)
plt.axvline(0, linewidth=0.8)
plt.xlabel("z")
plt.ylabel("sigmoid(z)")
plt.title("Función Sigmoid")
plt.show()
````

### Explora

Predice aproximadamente:

\[
\sigma(-5),\qquad \sigma(0),\qquad \sigma(5).
\]

Luego ejecuta:


````python
for value in [-5, 0, 5]:
    print(f"sigmoid({value}) = {sigmoid(value):.4f}")
````

<details>
<summary><strong>Interpretación</strong></summary>

- valores muy negativos → cerca de 0;
- \(z=0\) → 0.5;
- valores muy positivos → cerca de 1.

Por eso Sigmoid puede interpretarse como una salida parecida a una probabilidad en clasificación binaria.

</details>


## 4. Tanh

Otra función clásica es:

\[
\tanh(z).
\]

Su rango es:

\[
-1 \le \tanh(z) \le 1.
\]


````python
z = np.linspace(-5, 5, 300)

plt.figure(figsize=(7, 4))
plt.plot(z, np.tanh(z))
plt.axhline(0, linewidth=0.8)
plt.axvline(0, linewidth=0.8)
plt.xlabel("z")
plt.ylabel("tanh(z)")
plt.title("Función Tanh")
plt.show()
````

## 5. Comparación visual

Veamos las tres juntas.


````python
z = np.linspace(-5, 5, 300)

plt.figure(figsize=(8, 5))
plt.plot(z, relu(z), label="ReLU")
plt.plot(z, sigmoid(z), label="Sigmoid")
plt.plot(z, np.tanh(z), label="Tanh")
plt.xlabel("z")
plt.ylabel("activation(z)")
plt.title("Comparación de funciones de activación")
plt.legend()
plt.show()
````

## 6. Aplicando una activación a una neurona

Primero:

\[
z=\mathbf{w}\cdot\mathbf{x}+b.
\]

Luego:

\[
a=\mathrm{ReLU}(z).
\]


````python
x = np.array([0.8, 0.6, 0.2])
w = np.array([0.5, -0.3, 0.8])
b = -0.7

z = np.dot(w, x) + b
a = relu(z)

print("z =", z)
print("ReLU(z) =", a)
````

### Reto

Cambia solamente `b` hasta lograr que la salida de ReLU sea positiva.

<details>
<summary><strong>Pista</strong></summary>

Necesitas que:

\[
z>0.
\]

</details>

<details>
<summary><strong>Mostrar una posible solución</strong></summary>

Por ejemplo:

```python
b = 0.1
```

produce un valor de \(z\) positivo con estas entradas y pesos.

</details>


## 7. Ejemplo científico: decisión binaria

Imaginemos que una neurona produce:

\[
z=1.8.
\]

Usamos Sigmoid:


````python
z = 1.8
probability = sigmoid(z)

print(f"Salida sigmoid = {probability:.3f}")
````

Podemos interpretar esa salida como una señal entre 0 y 1.

Por ejemplo:

```text
0.00 → muy poca evidencia
0.50 → decisión incierta
1.00 → evidencia fuerte
```

En un problema meteorológico binario, podría representar algo como:

```text
probabilidad estimada de lluvia
```

aunque en un modelo real debemos calibrar e interpretar estas probabilidades con cuidado.


## 8. ¿Cuál activación usamos?

Una guía inicial:

| Lugar en la red | Activación común |
|---|---|
| Hidden layers | ReLU |
| Salida binaria | Sigmoid |
| Salida multiclase | Softmax |

No es una regla absoluta, pero es una buena primera referencia.

Más adelante veremos **Softmax** cuando construyamos redes de clasificación multiclase.
