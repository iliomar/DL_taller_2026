# Loss Functions

Ya podemos realizar un forward pass y obtener una predicción.

Ahora necesitamos responder:

> **¿Qué tan equivocada está la red?**

Una **loss function** convierte el error del modelo en un número.

```text
Prediction ──┐
             ├──► Loss
True value ──┘
```

- relacionar una loss pequeña con una mejor predicción.


````python
import numpy as np
import matplotlib.pyplot as plt
````

## 1. Un ejemplo de regresión

Supongamos que queremos predecir una temperatura.

Valor real:

\[
y=30
\]

Predicción:

\[
\hat{y}=26.
\]

Un error sencillo sería:

\[
y-\hat{y}=4.
\]

Pero normalmente queremos una cantidad siempre positiva.

Podemos usar:

\[
(y-\hat{y})^2.
\]


````python
y_true = 30
y_pred = 26

squared_error = (y_true - y_pred)**2

print("Squared error =", squared_error)
````

## 2. Mean Squared Error

Para muchos samples usamos:

\[
\mathrm{MSE}
=
\frac{1}{N}
\sum_{i=1}^{N}
(y_i-\hat{y}_i)^2.
\]


````python
y_true = np.array([30, 28, 31, 29])
y_pred = np.array([26, 29, 30, 27])

mse = np.mean((y_true - y_pred)**2)

print("MSE =", mse)
````

## 3. Visualizando MSE

Veamos cómo cambia el error cuando una predicción se aleja del valor real.


````python
y_true = 0.0
predictions = np.linspace(-4, 4, 300)
losses = (predictions - y_true)**2

plt.figure(figsize=(7, 4))
plt.plot(predictions, losses)
plt.xlabel("Predicción")
plt.ylabel("Squared Error")
plt.title("Squared Error para y = 0")
plt.show()
````

La curva es pequeña cerca de la respuesta correcta y aumenta rápidamente cuando nos alejamos.


## 4. Clasificación binaria

En clasificación binaria una red puede producir una salida como:

```text
0.95
```

si cree fuertemente en la clase 1.

Una loss muy utilizada es **Binary Cross Entropy (BCE)**:

\[
L
=
-\left[
y\log(\hat{y})
+
(1-y)\log(1-\hat{y})
\right].
\]


````python
def binary_cross_entropy(y_true, y_pred, eps=1e-8):
    y_pred = np.clip(y_pred, eps, 1-eps)
    return -(y_true*np.log(y_pred) + (1-y_true)*np.log(1-y_pred))
````

Supongamos que la respuesta correcta es:

\[
y=1.
\]

Comparemos varias predicciones.


````python
for prediction in [0.99, 0.90, 0.70, 0.50, 0.20, 0.01]:
    loss = binary_cross_entropy(1, prediction)
    print(f"prediction = {prediction:.2f} -> loss = {loss:.4f}")
````

## 5. Comparación gráfica de BCE


````python
pred = np.linspace(0.001, 0.999, 300)

loss_y1 = [binary_cross_entropy(1, p) for p in pred]
loss_y0 = [binary_cross_entropy(0, p) for p in pred]

plt.figure(figsize=(7, 4))
plt.plot(pred, loss_y1, label="y = 1")
plt.plot(pred, loss_y0, label="y = 0")
plt.xlabel("Predicción")
plt.ylabel("Binary Cross Entropy")
plt.title("Binary Cross Entropy")
plt.legend()
plt.show()
````

## 6. ¿Cuál loss usamos?

Una guía inicial:

| Problema | Loss común |
|---|---|
| Regresión | Mean Squared Error |
| Clasificación binaria | Binary Cross Entropy |
| Clasificación multiclase | Cross Entropy |

La elección depende del problema y de cómo representamos la salida.

## 8. La loss no es lo mismo que accuracy

Podemos tener:

```text
prediction 0.51 → clase correcta
prediction 0.99 → clase correcta
```

Ambas tienen la misma accuracy para ese sample, pero no representan el mismo nivel de confianza.

La loss conserva más información sobre la calidad de la predicción.
