# Backpropagation

Ahora tenemos todas las piezas principales del entrenamiento:

```text
Forward Pass
    ↓
Prediction
    ↓
Loss
    ↓
?
    ↓
Update Parameters
```

La pieza que falta es **Backpropagation**.

Backpropagation calcula cómo cambia la loss cuando cambiamos cada parámetro de la red.

## Objetivos

Al finalizar podrás:

- entender backpropagation como aplicación repetida de la chain rule;
- calcular derivadas en una red muy pequeña;
- seguir un gradiente hacia atrás;
- entender por qué `loss.backward()` es tan importante;
- conectar backpropagation con gradient descent.


````python
import numpy as np
````

## 1. El problema

Consideremos una red extremadamente pequeña:

\[
z=wx+b
\]

y para simplificar:

\[
\hat{y}=z.
\]

Usaremos:

\[
L=(\hat{y}-y)^2.
\]

El flujo es:

```text
x
↓
w, b
↓
z = wx+b
↓
prediction
↓
loss
```

Queremos saber:

\[
\frac{\partial L}{\partial w}
\]

y:

\[
\frac{\partial L}{\partial b}.
\]


## 2. Forward pass numérico

Supongamos:

\[
x=2,\qquad
w=0.5,\qquad
b=0.1,\qquad
y=3.
\]


````python
x = 2.0
w = 0.5
b = 0.1
y = 3.0

z = w*x + b
y_hat = z
loss = (y_hat - y)**2

print("z =", z)
print("prediction =", y_hat)
print("loss =", loss)
````

## 3. Chain Rule

La loss depende de \(w\) indirectamente:

\[
w
\rightarrow
z
\rightarrow
\hat{y}
\rightarrow
L.
\]

Por la chain rule:

\[
\frac{\partial L}{\partial w}
=
\frac{\partial L}{\partial \hat{y}}
\frac{\partial \hat{y}}{\partial z}
\frac{\partial z}{\partial w}.
\]


### Primer término

\[
L=(\hat{y}-y)^2
\]

por lo tanto:

\[
\frac{\partial L}{\partial \hat{y}}
=
2(\hat{y}-y).
\]


````python
dL_dyhat = 2*(y_hat - y)
print(dL_dyhat)
````

### Segundo término

Como:

\[
\hat{y}=z
\]

entonces:

\[
\frac{\partial \hat{y}}{\partial z}=1.
\]


### Tercer término

Como:

\[
z=wx+b
\]

entonces:

\[
\frac{\partial z}{\partial w}=x.
\]

Por lo tanto:


````python
dyhat_dz = 1.0
dz_dw = x

dL_dw = dL_dyhat * dyhat_dz * dz_dw

print("dL/dw =", dL_dw)
````

## 4. Gradiente respecto al bias

Como:

\[
\frac{\partial z}{\partial b}=1,
\]

entonces:

\[
\frac{\partial L}{\partial b}
=
\frac{\partial L}{\partial \hat{y}}
\frac{\partial \hat{y}}{\partial z}
\frac{\partial z}{\partial b}.
\]


````python
dz_db = 1.0
dL_db = dL_dyhat * dyhat_dz * dz_db

print("dL/db =", dL_db)
````

## 5. Actualizamos los parámetros

Ahora usamos Gradient Descent:

\[
w\leftarrow w-\eta\frac{\partial L}{\partial w}
\]

\[
b\leftarrow b-\eta\frac{\partial L}{\partial b}.
\]


````python
learning_rate = 0.05

w_new = w - learning_rate*dL_dw
b_new = b - learning_rate*dL_db

print("old w =", w, "new w =", w_new)
print("old b =", b, "new b =", b_new)
````

### Pregunta

Si volvemos a hacer el forward pass con los nuevos parámetros, ¿esperas una loss mayor o menor?

<details>
<summary><strong>Pista</strong></summary>

Gradient Descent intenta moverse en la dirección que reduce la loss.

</details>


````python
z_new = w_new*x + b_new
loss_new = (z_new - y)**2

print("old loss =", loss)
print("new loss =", loss_new)
````

<details>
<summary><strong>Mostrar solución</strong></summary>

La nueva loss es menor.

Eso significa que el update movió los parámetros en una dirección útil.

</details>


## 6. ¿Qué significa "backpropagation"?

El forward pass va:

```text
x → z → prediction → loss
```

Backpropagation recorre el cálculo en sentido contrario:

```text
loss
  ↓
gradient prediction
  ↓
gradient z
  ↓
gradient w, b
```

No estamos enviando datos "hacia atrás".

Estamos propagando **derivadas**.


## 7. ReLU y su derivada

Para ReLU:

\[
\mathrm{ReLU}(z)=\max(0,z).
\]

Su derivada es:

\[
\frac{d}{dz}\mathrm{ReLU}(z)
=
\begin{cases}
0 & z<0 \\
1 & z>0
\end{cases}
\]

Esto significa que el gradiente puede pasar o bloquearse dependiendo del valor de \(z\).


````python
def relu(z):
    return np.maximum(0, z)

def relu_grad(z):
    return 1.0 if z > 0 else 0.0

for z in [-2, -0.5, 0.5, 3]:
    print(z, "->", relu_grad(z))
````

## 9. PyTorch hace esto automáticamente

En PyTorch, el sistema de automatic differentiation construye un grafo de operaciones.

Después podemos usar:

```python
loss.backward()
```

para calcular gradientes.

Más adelante trabajaremos directamente con:

```python
optimizer.zero_grad()
loss.backward()
optimizer.step()
```

Estas tres líneas representan:

```text
limpiar gradientes
calcular gradientes
actualizar parámetros
```
