# Forward Pass

En los módulos anteriores construimos una red neuronal a partir de:

- features;
- weights;
- bias;
- funciones de activación;
- capas.

Ahora estudiaremos cómo **fluye la información a través de la red**.

A este proceso lo llamamos **forward pass**.

La idea general es:

```text
Features
   ↓
Layer 1
   ↓
Activation
   ↓
Layer 2
   ↓
Activation / Output
   ↓
Prediction
```

## 1. Recordatorio: una neurona

Una neurona calcula:

\[
z = \mathbf{w}\cdot\mathbf{x} + b
\]

y después:

\[
a = f(z).
\]

Para una red con varias capas repetimos esta operación varias veces.

Por ejemplo:

\[
\mathbf{z}^{(1)} = W^{(1)}\mathbf{x}+\mathbf{b}^{(1)}
\]

\[
\mathbf{a}^{(1)} = f(\mathbf{z}^{(1)})
\]

\[
\mathbf{z}^{(2)} = W^{(2)}\mathbf{a}^{(1)}+\mathbf{b}^{(2)}.
\]


## 2. Una red científica sencilla

Supongamos que queremos clasificar una señal astronómica usando tres features:

- brillo relativo;
- índice de color;
- variabilidad.

Arquitectura:

```text
3 inputs → 4 hidden neurons → 1 output
```

o de forma compacta:

```text
3 → 4 → 1
```


````python
x = np.array([0.80, 0.35, 0.60])

W1 = np.array([
    [ 0.50, -0.20,  0.80],
    [-0.30,  0.90,  0.10],
    [ 0.70,  0.40, -0.50],
    [ 0.20, -0.60,  1.00]
])

b1 = np.array([0.10, -0.20, 0.05, 0.30])

W2 = np.array([[0.60, -0.40, 0.90, 0.30]])
b2 = np.array([-0.20])
````

## 3. Primera capa

Calculamos:

\[
\mathbf{z}_1=W_1\mathbf{x}+\mathbf{b}_1.
\]


````python
z1 = W1 @ x + b1
print("z1 =", z1)
````

Ahora aplicamos ReLU:

\[
\mathbf{a}_1=\mathrm{ReLU}(\mathbf{z}_1).
\]


````python
def relu(z):
    return np.maximum(0, z)

a1 = relu(z1)

print("a1 =", a1)
````

### Pregunta

¿Cuál es la diferencia entre `z1` y `a1`?

<details>
<summary><strong>Pista</strong></summary>

Uno es el resultado antes de la activación y el otro después.

</details>

<details>
<summary><strong>Mostrar solución</strong></summary>

- `z1` es la combinación lineal \(W_1x+b_1\).
- `a1` es la salida después de aplicar ReLU.

```text
z1 → activation → a1
```

</details>


## 4. Output layer

La segunda capa recibe `a1`.

Calculamos:

\[
z_2=W_2a_1+b_2.
\]


````python
z2 = W2 @ a1 + b2
print("z2 =", z2)
````

Para una clasificación binaria podemos aplicar Sigmoid:

\[
\hat{y}=\sigma(z_2).
\]


````python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

y_hat = sigmoid(z2)

print("y_hat =", y_hat)
````

### Interpretación

La salida está entre 0 y 1.

Si definimos:

```text
y_hat > 0.5 → clase 1
y_hat ≤ 0.5 → clase 0
```

podemos convertir la salida continua en una decisión.


````python
prediction = 1 if y_hat.item() > 0.5 else 0

print("prediction =", prediction)
````
