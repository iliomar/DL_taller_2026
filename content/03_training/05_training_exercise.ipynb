## 6. Deriva los gradientes

Nuestro modelo es:

\[
\hat{y}_i=wx_i+b.
\]

La loss es:

\[
L=
\frac{1}{N}\sum_i(\hat{y}_i-y_i)^2.
\]

Los gradientes son:

\[
\frac{\partial L}{\partial w}
=
\frac{2}{N}
\sum_i
(\hat{y}_i-y_i)x_i

\[

\frac{2}{N}
\sum_i
(\hat{y}_i-y_i).
\]

<details>
<summary><strong>Pista</strong></summary>

Define primero:

```python

```

</details>

<details>
<summary><strong>Mostrar solución</strong></summary>

```python
def gradients(x, y_true, y_pred):
    error = y_pred - y_true

    dw = 2*np.mean(error*x)
    db = 2*np.mean(error)

    return dw, db
```

</details>

# El training loop completo

Todo el entrenamiento puede resumirse como:

```text
                TRAINING LOOP

Features
   ↓
Forward Pass
   ↓
Prediction
   ↓
Loss
   ↓
Backpropagation
   ↓
Gradients
   ↓
Update Parameters
   ↓
Repeat
```

Este ciclo es la base del entrenamiento de redes neuronales modernas.

La diferencia en una red profunda es que habrá muchos más weights, biases, capas y operaciones.
