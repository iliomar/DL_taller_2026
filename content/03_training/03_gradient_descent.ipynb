# Gradient Descent

Ya sabemos:

```text
Forward Pass
    ↓
Prediction
    ↓
Loss
```

Ahora necesitamos reducir esa loss.

La idea central de **Gradient Descent** es:

> mover los parámetros en la dirección que reduce el error.

## Objetivos

Al finalizar podrás:

- entender una loss como función de los parámetros;
- interpretar una pendiente o gradiente;
- actualizar un parámetro usando gradient descent;
- entender el papel del learning rate;
- visualizar convergencia;
- reconocer problemas con learning rates muy pequeños o muy grandes.


````python
import numpy as np
import matplotlib.pyplot as plt
````

## 1. Una loss muy sencilla

Consideremos:

\[
L(w)=(w-3)^2.
\]

El mínimo ocurre en:

\[
w=3.
\]

Imaginemos que el modelo todavía no sabe eso.

Comenzamos con:

\[
w=0.
\]


````python
def loss(w):
    return (w - 3)**2

w_values = np.linspace(-2, 8, 300)

plt.figure(figsize=(7, 4))
plt.plot(w_values, loss(w_values))
plt.xlabel("w")
plt.ylabel("Loss")
plt.title("Loss como función de un parámetro")
plt.show()
````

## 2. El gradiente

La derivada es:

\[
\frac{dL}{dw}=2(w-3).
\]

Esta cantidad nos dice la pendiente local de la loss.


````python
def gradient(w):
    return 2*(w - 3)

for w in [0, 2, 3, 5]:
    print(f"w = {w:>2} -> gradient = {gradient(w):>5.2f}")
````

### Pregunta

¿Qué ocurre con el gradiente en el mínimo?

<details>
<summary><strong>Mostrar solución</strong></summary>

En \(w=3\):

\[
\frac{dL}{dw}=0.
\]

En un mínimo suave, el gradiente es cero.

</details>


## 3. La regla de actualización

Gradient Descent utiliza:

\[
w_{\text{nuevo}}
=
w_{\text{viejo}}
-
\eta
\frac{dL}{dw}
\]

donde:

\[
\eta
\]

es el **learning rate**.


````python
w = 0.0
learning_rate = 0.1

for step in range(10):
    grad = gradient(w)
    w = w - learning_rate * grad
    print(f"step={step:2d}  w={w:.4f}  loss={loss(w):.6f}")
````

Observa cómo `w` se mueve gradualmente hacia 3.


## 4. Visualizando el entrenamiento


````python
w = 0.0
learning_rate = 0.1

history_w = [w]
history_loss = [loss(w)]

for step in range(25):
    w = w - learning_rate * gradient(w)
    history_w.append(w)
    history_loss.append(loss(w))

plt.figure(figsize=(7, 4))
plt.plot(history_loss, marker="o")
plt.xlabel("Step")
plt.ylabel("Loss")
plt.title("Loss durante Gradient Descent")
plt.show()
````

## 5. Learning rate

El learning rate controla el tamaño del paso.

### Muy pequeño

```text
aprendizaje muy lento
```

### Apropiado

```text
convergencia estable
```

### Muy grande

```text
puede saltar sobre el mínimo o divergir
```


````python
def run_gradient_descent(lr, steps=15):
    w = 0.0
    losses = []

    for _ in range(steps):
        losses.append(loss(w))
        w = w - lr * gradient(w)

    return w, losses

for lr in [0.01, 0.1, 0.9, 1.1]:
    final_w, losses = run_gradient_descent(lr)
    print(f"lr={lr:>4} -> final w={final_w:.4f}, final loss={losses[-1]:.6f}")
````

### Pregunta

¿Cuál de estos learning rates parece más estable?

<details>
<summary><strong>Pista</strong></summary>

Busca una loss que disminuya sin oscilar descontroladamente.

</details>

<details>
<summary><strong>Mostrar solución</strong></summary>

`0.1` es estable para este ejemplo.

`0.01` también es estable, pero más lento.

Valores demasiado grandes pueden oscilar o divergir.

</details>

## 7. Conexión con una red neuronal

Una red real tiene muchos parámetros:

```text
w1
w2
w3
...
b1
b2
...
```

Gradient Descent actualiza cada uno:

\[
\theta
\leftarrow
\theta
-
\eta
\frac{\partial L}{\partial \theta}.
\]

Aquí \(\theta\) representa cualquier parámetro entrenable.
