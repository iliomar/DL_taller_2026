## Contexto del dataset Iris

El dataset Iris contiene medidas de flores de tres especies:

- Setosa
- Versicolor
- Virginica

Para cada flor se midieron cuatro características:

- Longitud del sépalo
- Ancho del sépalo
- Longitud del pétalo
- Ancho del pétalo

La pregunta del proyecto es:

> ¿Puede un modelo de Machine Learning aprender a identificar la especie de una flor usando estas medidas?

Este dataset viene incluido en `scikit-learn`, por lo que no necesitamos descargar archivos externos.

## Parte 1: Importar librerías

Primero importamos las herramientas que vamos a usar.

Ejecuta la siguiente celda.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, ConfusionMatrixDisplay, classification_report

print("Librerías importadas correctamente.")
```

## Parte 2: Cargar los datos reales

Ahora cargamos el dataset Iris desde `scikit-learn`.

Ejecuta la celda y observa qué contiene el objeto `iris`.

```python
iris = load_iris()

iris.keys()
```

```python
print("Nombres de las variables:")
print(iris.feature_names)

print("\nNombres de las especies:")
print(iris.target_names)

print("\nTamaño de los datos:")
print(iris.data.shape)
```

## Parte 3: Crear un DataFrame

Un `DataFrame` es una tabla. Esto hace que los datos sean más fáciles de leer, explorar y explicar a los estudiantes.

```python
df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

df["species_id"] = iris.target
df["species"] = df["species_id"].map({
    0: "setosa",
    1: "versicolor",
    2: "virginica"
})

df.head()
```

## Parte 5: Visualizar los datos

Las gráficas nos ayudan a ver patrones. Primero miraremos la relación entre longitud del pétalo y ancho del pétalo.

```python
plt.figure(figsize=(8, 5))

for species in df["species"].unique():
    subset = df[df["species"] == species]
    plt.scatter(
        subset["petal length (cm)"],
        subset["petal width (cm)"],
        label=species,
        alpha=0.8
    )

plt.xlabel("Longitud del pétalo (cm)")
plt.ylabel("Ancho del pétalo (cm)")
plt.title("Relación entre longitud y ancho del pétalo")
plt.legend()
plt.show()
```

## Parte 6: Preparar `X` y `y`

En Machine Learning supervisado necesitamos separar los datos en dos partes:

- `X`: variables de entrada o características.
- `y`: respuesta correcta o variable objetivo.

En este caso, `X` serán las medidas de las flores y `y` será la especie de la flor.

```python
X = df[[
    "sepal length (cm)",
    "sepal width (cm)",
    "petal length (cm)",
    "petal width (cm)"
]]

y = df["species"]

print("Primeras filas de X:")
display(X.head())

print("Primeras filas de y:")
display(y.head())
```

## Parte 7: Dividir datos en entrenamiento y prueba

El modelo debe aprender con una parte de los datos y evaluarse con otra parte que no haya visto. Esto nos ayuda a saber si el modelo realmente aprendió patrones generales.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

print("Datos para entrenar:", X_train.shape)
print("Datos para probar:", X_test.shape)
```

### Preguntas para discutir

## Parte 8: Entrenar el primer modelo

Usaremos **Regresión Logística** como primer modelo de clasificación. Aunque tiene la palabra "regresión" en el nombre, este modelo se puede usar para clasificar categorías.

```python
model = LogisticRegression(max_iter=1000)

model.fit(X_train, y_train)

print("Modelo entrenado correctamente.")
```

### Pregunta clave

## Parte 9: Hacer predicciones

Ahora el modelo intentará predecir la especie de las flores del conjunto de prueba.

```python
y_pred = model.predict(X_test)

comparacion = pd.DataFrame({
    "Real": y_test.values,
    "Predicción": y_pred
})

comparacion.head(15)
```

### Actividad

## Parte 10: Evaluar con accuracy

El `accuracy` nos dice qué proporción de predicciones fueron correctas.

```python
accuracy = accuracy_score(y_test, y_pred)

print(f"Accuracy del modelo: {accuracy:.2f}")
```

### Interpretación

- `1.00` significa que el modelo acertó todas las predicciones.
- `0.80` significa que acertó el 80%.
- `0.50` significa que acertó la mitad.

## Parte 11: Matriz de confusión

La matriz de confusión muestra dónde el modelo acierta y dónde se equivoca. Es una herramienta muy útil porque no solo nos dice cuántos errores hubo, sino también qué clases se confundieron.

```python
ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred
)

plt.title("Matriz de confusión - Regresión Logística")
plt.show()
```

```python
print(classification_report(y_test, y_pred))
```

# Parte 13: Retos guiados

## Reto 1: Usar solo variables del pétalo

Copia y ejecuta la siguiente celda. Luego compara el accuracy con el modelo anterior.

```python
X_petal = df[[
    "petal length (cm)",
    "petal width (cm)"
]]

y = df["species"]

X_train_p, X_test_p, y_train_p, y_test_p = train_test_split(
    X_petal,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

model_petal = LogisticRegression(max_iter=1000)
model_petal.fit(X_train_p, y_train_p)

y_pred_p = model_petal.predict(X_test_p)

accuracy_petal = accuracy_score(y_test_p, y_pred_p)

print(f"Accuracy usando solo pétalos: {accuracy_petal:.2f}")
```

### Preguntas del Reto 1

1. ¿El modelo mejoró, empeoró o quedó parecido?

## Reto 2: Usar solo variables del sépalo

Ahora probaremos con las variables del sépalo.

```python
X_sepal = df[[
    "sepal length (cm)",
    "sepal width (cm)"
]]

y = df["species"]

X_train_s, X_test_s, y_train_s, y_test_s = train_test_split(
    X_sepal,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

model_sepal = LogisticRegression(max_iter=1000)
model_sepal.fit(X_train_s, y_train_s)

y_pred_s = model_sepal.predict(X_test_s)

accuracy_sepal = accuracy_score(y_test_s, y_pred_s)

print(f"Accuracy usando solo sépalos: {accuracy_sepal:.2f}")
```

## Reto 4: Probar otro modelo

Ahora probaremos un modelo llamado **Árbol de Decisión**. Este modelo toma decisiones usando preguntas como: ¿la longitud del pétalo es mayor que cierto valor?

```python
tree_model = DecisionTreeClassifier(
    random_state=42,
    max_depth=3
)

tree_model.fit(X_train, y_train)

y_pred_tree = tree_model.predict(X_test)

accuracy_tree = accuracy_score(y_test, y_pred_tree)

print(f"Accuracy del Árbol de Decisión: {accuracy_tree:.2f}")
```

```python
ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred_tree
)

plt.title("Matriz de confusión - Árbol de Decisión")
plt.show()
```
