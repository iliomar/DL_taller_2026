# Paso 1: Importar librerías

# Paso 2: Cargar la data

Ejecuta la celda.  
El dataset se cargará desde UCI Machine Learning Repository.


````python
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv"

df = pd.read_csv(url, sep=";")

df.head()
````

# Paso 3: Explorar rápidamente la data

# Paso 4: Crear la variable objetivo

La columna original `quality` va de valores bajos a altos.

Para convertir esto en clasificación binaria, crearemos:

```python
good_quality = 1 si quality >= 6
good_quality = 0 si quality < 6
```

Ejecuta la celda.


````python
df["good_quality"] = (df["quality"] >= 6).astype(int)

df[["quality", "good_quality"]].head(10)
````

# Paso 5: Visualizar la variable objetivo

````python
df["good_quality"].value_counts().plot(kind="bar")

plt.title("Distribución de calidad del vino")
plt.xlabel("0 = regular, 1 = buena calidad")
plt.ylabel("Cantidad")
plt.show()
````

# Paso 6: Visualizar alcohol vs calidad

Observa si el alcohol parece estar relacionado con la calidad.


````python
plt.figure(figsize=(8,5))
plt.scatter(df["alcohol"], df["good_quality"], alpha=0.2)
plt.xlabel("Alcohol")
plt.ylabel("Buena calidad")
plt.title("Alcohol vs calidad")
plt.show()
````
