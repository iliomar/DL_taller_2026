## 1. La pregunta científica

Tenemos mediciones de flores pertenecientes a tres especies:

- *Iris setosa*
- *Iris versicolor*
- *Iris virginica*

Queremos investigar:

> **¿Podemos distinguir la especie de una flor utilizando solamente mediciones de sus sépalos y pétalos?**


## 2. Importamos las herramientas


````python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
````

## 3. Cargamos las observaciones


````python
iris = load_iris()

print('Features disponibles:')
for feature in iris.feature_names:
    print('-', feature)

print('\nEspecies:')
for species in iris.target_names:
    print('-', species)
````

Las features son cuatro mediciones físicas:

1. largo del sépalo,
2. ancho del sépalo,
3. largo del pétalo,
4. ancho del pétalo.

La **label** es la especie de la flor.


## 4. Construimos una tabla


````python
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species_id'] = iris.target
df['species'] = df['species_id'].map(dict(enumerate(iris.target_names)))
df.head()
````

## 5. Samples y features

Representamos las mediciones con `X` y las especies con `y`.


````python
X = iris.data
y = iris.target

print('X.shape =', X.shape)
print('y.shape =', y.shape)
````

Interpreta el resultado:

```text
Número de observaciones = ______
Número de features      = ______
Número de labels        = ______
```

Recuerda:

$$
X.shape = (n_{samples}, n_{features})
$$


## 6. Inspeccionamos una observación


````python
sample = 25

print('Observación', sample)
for name, value in zip(iris.feature_names, X[sample]):
    print(f'{name}: {value:.2f} cm')

print('\nEspecie:', iris.target_names[y[sample]])
````

Una observación individual puede pensarse como:

```text
4 mediciones físicas ───► modelo ───► especie
```


## 7. ¿Cómo varía una medición?

Visualicemos el largo del pétalo para todas las flores.


````python
plt.figure(figsize=(7, 4))
plt.hist(df['petal length (cm)'], bins=15, edgecolor='black')
plt.xlabel('Largo del pétalo (cm)')
plt.ylabel('Número de flores')
plt.title('Distribución del largo del pétalo')
plt.show()
````

### Preguntas

- ¿Las flores tienen todas el mismo largo de pétalo?
- ¿Qué rango aproximado observas?
- ¿Por qué una variable que cambia entre especies podría ser útil para clasificarlas?


## 8. Dos mediciones a la vez

Ahora compararemos el largo y el ancho del pétalo.


````python
plt.figure(figsize=(7, 5))

for class_id, species in enumerate(iris.target_names):
    mask = y == class_id
    plt.scatter(X[mask, 2], X[mask, 3], label=species, alpha=0.75)

plt.xlabel('Largo del pétalo (cm)')
plt.ylabel('Ancho del pétalo (cm)')
plt.title('Características del pétalo por especie')
plt.legend()
plt.show()
````

### Interpreta el gráfico

1. ¿Qué especie parece más fácil de distinguir?
2. ¿Qué especies presentan mayor solapamiento?
3. ¿Parece que estas dos features contienen información sobre la especie?

El objetivo del modelo será aprender este tipo de patrones automáticamente.


## 9. ¿Son todas las mediciones igual de informativas?

Comparemos ahora largo y ancho del sépalo.


````python
plt.figure(figsize=(7, 5))

for class_id, species in enumerate(iris.target_names):
    mask = y == class_id
    plt.scatter(X[mask, 0], X[mask, 1], label=species, alpha=0.75)

plt.xlabel('Largo del sépalo (cm)')
plt.ylabel('Ancho del sépalo (cm)')
plt.title('Características del sépalo por especie')
plt.legend()
plt.show()
````

### Comparación científica

Compara ambos gráficos.

- ¿Qué par de mediciones separa mejor las especies?
- ¿Qué podría indicar esto sobre la utilidad de diferentes features?

> **No todas las mediciones contienen la misma cantidad de información para una tarea determinada.**
